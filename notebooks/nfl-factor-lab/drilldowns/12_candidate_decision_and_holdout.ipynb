# 12_candidate_decision_and_holdout

只有完成专家审核并冻结 shortlist 后，才允许解锁一次 historical holdout。默认不运行。

```python
from prediction_market.sports.nfl_factor_lab import (
    freeze_factor_shortlist,
    run_factor_discovery,
    run_locked_historical_holdout,
    verify_and_load_x13_inputs,
)

claim_boundary = "PRELIMINARY_SOURCE_TIME_ONLY"
verified_inputs = verify_and_load_x13_inputs()
human_reviews = ()
shortlist_lock = None
```

```python
if human_reviews:
    discovery = run_factor_discovery(verified_inputs)
    shortlist_lock = freeze_factor_shortlist(verified_inputs, discovery, human_reviews)

holdout = None
if shortlist_lock is not None:
    holdout = run_locked_historical_holdout(verified_inputs, shortlist_lock)

display({"claim_boundary": claim_boundary, "shortlist_lock": shortlist_lock, "holdout": holdout})
```
